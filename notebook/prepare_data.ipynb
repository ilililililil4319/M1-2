```python
import os
from dotenv import load_dotenv
import opendartreader
import pandas as pd

load_dotenv('../backend/.env')
DART_API_KEY = os.getenv('DART_API_KEY')

dart = opendartreader.OpenDartReader(DART_API_KEY)
print("연결 확인:", dart.company('035420')['corp_name'])
```

```python
CORP = '035420'  # 네이버
YEARS = range(2020, 2026)
REPRT_CODES = {'11013': 'Q1', '11012': 'Q2누적', '11014': 'Q3누적', '11011': '연간(사업보고서)'}

records = []
for year in YEARS:
    for code, label in REPRT_CODES.items():
        try:
            fs = dart.finstate(CORP, year, reprt_code=code)
            if fs is not None and len(fs) > 0:
                fs = fs.copy()
                fs['bsns_year'] = year
                fs['reprt_code'] = code
                fs['reprt_label'] = label
                records.append(fs)
        except Exception as e:
            print(f"{year} {label} 수집 실패:", e)

raw = pd.concat(records, ignore_index=True)
raw.to_csv('../data/naver_financials_raw.csv', index=False, encoding='utf-8-sig')
print("원본 데이터 shape:", raw.shape)
raw[['bsns_year', 'reprt_label', 'account_nm', 'thstrm_amount']].head(15)
```

```python
target_items = ['매출액', '영업이익', '당기순이익(손실)', '자산총계', '부채총계', '자본총계', '영업활동으로인한현금흐름']

filtered = raw[raw['account_nm'].isin(target_items)][
    ['bsns_year', 'reprt_label', 'account_nm', 'thstrm_amount']
]
print(filtered['account_nm'].unique())
filtered.head(20)
```

```python
# ① CFS(연결재무제표)만 필터링
cfs = raw[raw['fs_div'] == 'CFS'].copy()
print("CFS만 필터링:", cfs.shape)

# ② 현금흐름표는 별도 함수로 수집 (finstate_all 사용, 전체 재무제표 항목 포함)
cf_records = []
for year in YEARS:
    for code, label in REPRT_CODES.items():
        try:
            fs_all = dart.finstate_all(CORP, year, reprt_code=code, fs_div='CFS')
            if fs_all is not None and len(fs_all) > 0:
                fs_all = fs_all.copy()
                fs_all['bsns_year'] = year
                fs_all['reprt_code'] = code
                fs_all['reprt_label'] = label
                cf_records.append(fs_all)
        except Exception as e:
            print(f"{year} {label} 전체재무제표 수집 실패:", e)

full = pd.concat(cf_records, ignore_index=True)
print("전체 재무제표(finstate_all) shape:", full.shape)

# 현금흐름표 관련 계정명 찾기
cf_items = full[full['account_nm'].str.contains('현금흐름|영업활동', na=False)]
print(cf_items['account_nm'].unique())
```

```python
# ① 현금흐름 항목만 추출 (CFS만)
cf = full[(full['fs_div']=='CFS') & (full['account_nm']=='영업활동으로부터 창출된 현금흐름')][
    ['bsns_year','reprt_label','thstrm_amount']
].copy()
cf['account_nm'] = '영업활동현금흐름'

# ② BS+IS(CFS) + 현금흐름 합치기
combined = pd.concat([
    cfs[['bsns_year','reprt_label','account_nm','thstrm_amount']],
    cf[['bsns_year','reprt_label','account_nm','thstrm_amount']]
], ignore_index=True)

# ③ 금액을 숫자로 변환 (콤마 제거)
combined['thstrm_amount'] = combined['thstrm_amount'].astype(str).str.replace(',', '', regex=False)
combined['thstrm_amount'] = pd.to_numeric(combined['thstrm_amount'], errors='coerce')

# ④ wide 형태로 피벗 (분기 하나당 한 줄, 계정과목이 컬럼으로)
pivot = combined.pivot_table(
    index=['bsns_year','reprt_label'],
    columns='account_nm',
    values='thstrm_amount',
    aggfunc='first'
).reset_index()

print(pivot.shape)
pivot
```

```python
# ① 현금흐름표(CF)에서 영업활동현금흐름만 추출 (fs_div 필터 불필요 - 이미 CFS로 수집됨)
cf = full[(full['sj_div']=='CF') & (full['account_nm']=='영업활동으로부터 창출된 현금흐름')][
    ['bsns_year','reprt_label','thstrm_amount']
].copy()
cf['account_nm'] = '영업활동현금흐름'

# ② BS+IS(cfs, 이전 셀에서 이미 CFS로 필터링됨) + 현금흐름 합치기
combined = pd.concat([
    cfs[['bsns_year','reprt_label','account_nm','thstrm_amount']],
    cf[['bsns_year','reprt_label','account_nm','thstrm_amount']]
], ignore_index=True)

# ③ 금액을 숫자로 변환 (콤마 제거)
combined['thstrm_amount'] = combined['thstrm_amount'].astype(str).str.replace(',', '', regex=False)
combined['thstrm_amount'] = pd.to_numeric(combined['thstrm_amount'], errors='coerce')

# ④ wide 형태로 피벗
pivot = combined.pivot_table(
    index=['bsns_year','reprt_label'],
    columns='account_nm',
    values='thstrm_amount',
    aggfunc='first'
).reset_index()

print(pivot.shape)
pivot
```

```python
df = pivot.copy()

# 컬럼명 영문으로 정리 (다루기 쉽게)
df = df.rename(columns={
    '매출액':'revenue', '영업이익':'operating_profit', '당기순이익(손실)':'net_income',
    '자산총계':'total_assets', '부채총계':'total_liabilities', '자본총계':'total_equity',
    '유동자산':'current_assets', '유동부채':'current_liabilities',
    '영업활동현금흐름':'operating_cf'
})

# 계산 지표 추가
df['operating_margin'] = df['operating_profit'] / df['revenue'] * 100        # 영업이익률(%)
df['net_margin'] = df['net_income'] / df['revenue'] * 100                    # 순이익률(%)
df['debt_ratio'] = df['total_liabilities'] / df['total_equity'] * 100        # 부채비율(%)
df['current_ratio'] = df['current_assets'] / df['current_liabilities'] * 100 # 유동비율(%)
df['equity_ratio'] = df['total_equity'] / df['total_assets'] * 100           # 자기자본비율(%)

# 분기말 날짜 만들기 (quarter → 날짜)
quarter_end = {'Q1':'-03-31', 'Q2누적':'-06-30', 'Q3누적':'-09-30', '연간(사업보고서)':'-12-31'}
df['date'] = df['bsns_year'].astype(str) + df['reprt_label'].map(quarter_end)

# 매출증가율(YoY)은 같은 분기 전년 대비이므로, 정렬 후 계산
df = df.sort_values(['reprt_label','bsns_year'])
df['revenue_yoy'] = df.groupby('reprt_label')['revenue'].pct_change() * 100
df = df.sort_values('date').reset_index(drop=True)

print(df[['date','revenue','operating_profit','net_income','debt_ratio','current_ratio','equity_ratio','operating_cf','revenue_yoy']])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['font.family'] = 'Malgun Gothic'
mpl.rcParams['axes.unicode_minus'] = False

long_df = pd.read_csv('../data/naver_financials.csv')
wide = long_df.pivot_table(index='date', columns='memo', values='value', aggfunc='first').reset_index()
wide['date'] = pd.to_datetime(wide['date'])
wide = wide.sort_values('date').reset_index(drop=True)
wide.columns.name = None

wide['매출액_4Q이동평균'] = wide['매출액'].rolling(4).mean()
wide['영업이익_4Q이동평균'] = wide['영업이익'].rolling(4).mean()
wide['부채비율_변동성'] = wide['부채비율(%)'].rolling(4).std()

print(wide.shape)
wide[['date','매출액','매출액_4Q이동평균','영업이익','영업이익_4Q이동평균','부채비율(%)','부채비율_변동성']]
```

```python
import os
os.makedirs('../output', exist_ok=True)

fig, ax = plt.subplots(figsize=(10,5))
ax.plot(wide['date'], wide['매출액'], label='매출액', color='steelblue', alpha=0.4)
ax.plot(wide['date'], wide['매출액_4Q이동평균'], label='매출액 4분기 이동평균', color='steelblue', linewidth=2)
ax.plot(wide['date'], wide['영업이익'], label='영업이익', color='darkorange', alpha=0.4)
ax.plot(wide['date'], wide['영업이익_4Q이동평균'], label='영업이익 4분기 이동평균', color='darkorange', linewidth=2)
ax.set_title('네이버 매출·영업이익 추이 및 4분기 이동평균')
ax.set_ylabel('금액(원)')
ax.legend()
plt.tight_layout()
plt.savefig('../output/01_revenue_profit_trend.png', dpi=120)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(wide['date'], wide['부채비율(%)'], label='부채비율(%)', color='crimson')
ax.plot(wide['date'], wide['유동비율(%)'], label='유동비율(%)', color='seagreen')
ax.set_title('네이버 부채비율·유동비율 추이')
ax.set_ylabel('%')
ax.legend()
plt.tight_layout()
plt.savefig('../output/02_stability_ratios.png', dpi=120)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.bar(wide['date'], wide['영업활동현금흐름'], width=60, color='slateblue')
ax.axhline(0, color='gray', linewidth=0.8)
ax.set_title('네이버 영업활동현금흐름 추이')
ax.set_ylabel('금액(원)')
plt.tight_layout()
plt.savefig('../output/03_operating_cashflow.png', dpi=120)
plt.show()
```
